### Data structure

#### A data structure is a way of organizing, managing, and storing data in a computer so that it can be accessed and modified efficiently. Data structures form the foundation for creating efficient algorithms and solving computational problems. They can be classified into two main categories:

#### Linear Data Structures &  Non-Linear Data Structures



#### Linear Data Structures

Definition: Data is organized sequentially, and elements are accessed one after the other.

Examples:

Arrays: Fixed-size collections of elements stored in contiguous memory locations.

Linked Lists: Collections of nodes where each node points to the next, forming a chain.

Stacks: Last-In-First-Out (LIFO) structures.

Queues: First-In-First-Out (FIFO) structures.

```python
#Array

# Define an array (list in Python)
numbers = [10, 20, 30, 40, 50]

# Access elements
print("First element:", numbers[0])  # Output: 10
print("Last element:", numbers[-1])  # Output: 50

# Modify an element
numbers[2] = 35
print("Modified array:", numbers)  # Output: [10, 20, 35, 40, 50]

# Add elements
numbers.append(60)
print("After appending:", numbers)  # Output: [10, 20, 35, 40, 50, 60]

# Remove elements
numbers.pop(1)  # Removes the second element (20)
print("After removing second element:", numbers)  # Output: [10, 35, 40, 50, 60]

# Iterate through the array
print("Array elements:")
for number in numbers:
    print(number)

# Find the length of the array
print("Length of the array:", len(numbers))  # Output: 5
```

```python
#Linked kist

class Node:
    """A node in the linked list."""
    def __init__(self, data):
        self.data = data  # The data stored in the node
        self.next = None  # Pointer to the next node


class LinkedList:
    """A singly linked list."""
    def __init__(self):
        self.head = None  # Head of the linked list

    def append(self, data):
        """Add a node with the given data to the end of the linked list."""
        new_node = Node(data)
        if not self.head:  # If the list is empty
            self.head = new_node
            return
        last_node = self.head
        while last_node.next:  # Traverse to the end of the list
            last_node = last_node.next
        last_node.next = new_node

    def print_list(self):
        """Print the data in the linked list."""
        current_node = self.head
        while current_node:
            print(current_node.data, end=" -> ")
            current_node = current_node.next
        print("None")

    def delete(self, key):
        """Delete the first occurrence of a node with the given data."""
        current_node = self.head

        # If the node to be deleted is the head node
        if current_node and current_node.data == key:
            self.head = current_node.next
            current_node = None
            return

        # Search for the node to be deleted
        prev = None
        while current_node and current_node.data != key:
            prev = current_node
            current_node = current_node.next

        # If the data is not found
        if current_node is None:
            return

        # Remove the node
        prev.next = current_node.next
        current_node = None


# Example usage
linked_list = LinkedList()
linked_list.append(10)
linked_list.append(20)
linked_list.append(30)
linked_list.print_list()  # Output: 10 -> 20 -> 30 -> None

linked_list.delete(20)
linked_list.print_list()  # Output: 10 -> 30 -> None
```

```python
#Stacks

# Stack implementation using Python list
stack = []

# Push elements onto the stack
stack.append(10)
stack.append(20)
stack.append(30)

print("Stack after pushes:", stack)  # Output: [10, 20, 30]

# Pop elements from the stack
print("Popped element:", stack.pop())  # Output: 30
print("Stack after pop:", stack)       # Output: [10, 20]

# Peek at the top element
print("Top element:", stack[-1])       # Output: 20

# Check if the stack is empty
print("Is the stack empty?", len(stack) == 0)  # Output: False
```

```python
class Stack:
    def __init__(self):
        self.stack = []

    def push(self, item):
        """Add an item to the top of the stack."""
        self.stack.append(item)

    def pop(self):
        """Remove and return the top item from the stack."""
        if not self.is_empty():
            return self.stack.pop()
        else:
            return "Stack is empty"

    def peek(self):
        """Return the top item without removing it."""
        if not self.is_empty():
            return self.stack[-1]
        else:
            return "Stack is empty"

    def is_empty(self):
        """Check if the stack is empty."""
        return len(self.stack) == 0

    def size(self):
        """Return the number of items in the stack."""
        return len(self.stack)


# Example usage
stack = Stack()
stack.push(10)
stack.push(20)
stack.push(30)

print("Top element:", stack.peek())  # Output: 30
print("Popped element:", stack.pop())  # Output: 30
print("Is the stack empty?", stack.is_empty())  # Output: False
print("Stack size:", stack.size())  # Output: 2
```

```python
class Queue:
    def __init__(self):
        self.queue = []

    def enqueue(self, item):
        """Add an item to the end of the queue."""
        self.queue.append(item)

    def dequeue(self):
        """Remove and return the item from the front of the queue."""
        if not self.is_empty():
            return self.queue.pop(0)
        else:
            return "Queue is empty"

    def peek(self):
        """Return the front item without removing it."""
        if not self.is_empty():
            return self.queue[0]
        else:
            return "Queue is empty"

    def is_empty(self):
        """Check if the queue is empty."""
        return len(self.queue) == 0

    def size(self):
        """Return the number of items in the queue."""
        return len(self.queue)


# Example usage
queue = Queue()
queue.enqueue(10)
queue.enqueue(20)
queue.enqueue(30)

print("Front element:", queue.peek())  # Output: 10
print("Dequeued element:", queue.dequeue())  # Output: 10
print("Queue size:", queue.size())  # Output: 2
```

```python
#hash table 

class HashTable:
    def __init__(self, size):
        """Initialize the hash table with a fixed size."""
        self.size = size
        self.table = [[] for _ in range(size)]  # Create a list of empty lists (buckets)

    def _hash(self, key):
        """Generate a hash for the key."""
        return hash(key) % self.size

    def insert(self, key, value):
        """Insert a key-value pair into the hash table."""
        index = self._hash(key)
        for pair in self.table[index]:
            if pair[0] == key:  # If the key already exists, update its value
                pair[1] = value
                return
        self.table[index].append([key, value])  # Otherwise, add a new key-value pair

    def get(self, key):
        """Retrieve a value by its key."""
        index = self._hash(key)
        for pair in self.table[index]:
            if pair[0] == key:
                return pair[1]  # Return the value if the key is found
        return None  # Return None if the key is not found

    def delete(self, key):
        """Delete a key-value pair from the hash table."""
        index = self._hash(key)
        for pair in self.table[index]:
            if pair[0] == key:
                self.table[index].remove(pair)  # Remove the pair if the key is found
                return True
        return False  # Return False if the key is not found

    def display(self):
        """Print the hash table."""
        for i, bucket in enumerate(self.table):
            print(f"Bucket {i}: {bucket}")


# Example usage
hash_table = HashTable(5)  # Create a hash table with 5 buckets
hash_table.insert("name", "Alice")
hash_table.insert("age", 25)
hash_table.insert("city", "New York")

print("Get 'name':", hash_table.get("name"))  # Output: Alice
print("Get 'age':", hash_table.get("age"))    # Output: 25

hash_table.display()

hash_table.delete("age")
hash_table.display()
```

#### Non-Linear Data Structures

Definition: Data is not arranged in a sequential manner, allowing for more complex relationships.

Examples:

Trees: Hierarchical structures with a root and child nodes (e.g., binary trees, binary search trees).

Graphs: Consist of nodes (vertices) connected by edges, representing networks.

Heaps: Specialized tree-based structures used for priority management.

```python
# tree 

class Node:
    """Represents a node in the tree."""
    def __init__(self, value):
        self.value = value  # The data of the node
        self.left = None    # Pointer to the left child
        self.right = None   # Pointer to the right child


class BinaryTree:
    """Represents a binary tree."""
    def __init__(self):
        self.root = None

    def insert(self, value):
        """Inserts a value into the tree."""
        if self.root is None:
            self.root = Node(value)
        else:
            self._insert(self.root, value)

    def _insert(self, current_node, value):
        """Helper function to insert a value in the correct position."""
        if value < current_node.value:
            if current_node.left is None:
                current_node.left = Node(value)
            else:
                self._insert(current_node.left, value)
        else:
            if current_node.right is None:
                current_node.right = Node(value)
            else:
                self._insert(current_node.right, value)

    def inorder_traversal(self):
        """Performs in-order traversal of the tree."""
        result = []
        self._inorder_traversal(self.root, result)
        return result

    def _inorder_traversal(self, node, result):
        if node:
            self._inorder_traversal(node.left, result)
            result.append(node.value)
            self._inorder_traversal(node.right, result)

    def search(self, value):
        """Searches for a value in the tree."""
        return self._search(self.root, value)

    def _search(self, node, value):
        if not node:
            return False
        if node.value == value:
            return True
        elif value < node.value:
            return self._search(node.left, value)
        else:
            return self._search(node.right, value)


# Example usage
bt = BinaryTree()
bt.insert(50)
bt.insert(30)
bt.insert(70)
bt.insert(20)
bt.insert(40)
bt.insert(60)
bt.insert(80)

print("In-order Traversal:", bt.inorder_traversal())  # Output: [20, 30, 40, 50, 60, 70, 80]
print("Search for 60:", bt.search(60))  # Output: True
print("Search for 90:", bt.search(90))  # Output: False
```

```python
class Node:
    """Represents a node in the binary search tree."""
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    """Represents the binary search tree."""
    def __init__(self):
        self.root = None
    
    def insert(self, value):
        """Insert a value into the BST."""
        if not self.root:
            self.root = Node(value)
        else:
            self._insert(self.root, value)

    def _insert(self, node, value):
        if value < node.value:
            if node.left is None:
                node.left = Node(value)
            else:
                self._insert(node.left, value)
        elif value > node.value:
            if node.right is None:
                node.right = Node(value)
            else:
                self._insert(node.right, value)

    def search(self, value):
        """Search for a value in the BST."""
        return self._search(self.root, value)

    def _search(self, node, value):
        if node is None:
            return False
        if value == node.value:
            return True
        elif value < node.value:
            return self._search(node.left, value)
        else:
            return self._search(node.right, value)

    def inorder_traversal(self):
        """Return the elements of the BST in sorted order."""
        result = []
        self._inorder_traversal(self.root, result)
        return result

    def _inorder_traversal(self, node, result):
        if node:
            self._inorder_traversal(node.left, result)
            result.append(node.value)
            self._inorder_traversal(node.right, result)

# Example usage
bst = BinarySearchTree()
bst.insert(50)
bst.insert(30)
bst.insert(70)
bst.insert(20)
bst.insert(40)
bst.insert(60)
bst.insert(80)

print("In-order traversal:", bst.inorder_traversal())  # Output: [20, 30, 40, 50, 60, 70, 80]
print("Search for 60:", bst.search(60))  # Output: True
print("Search for 90:", bst.search(90))  # Output: False
```

```python
class MaxHeap:
    def __init__(self):
        self.heap = []

    def insert(self, value):
        """Insert a value into the heap and maintain heap property."""
        self.heap.append(value)  # Add the new value at the end of the list
        self._heapify_up(len(self.heap) - 1)  # Move the value up to maintain the max-heap property

    def _heapify_up(self, index):
        """Move the element at the given index up to restore heap property."""
        parent = (index - 1) // 2  # Find the parent index
        if index > 0 and self.heap[index] > self.heap[parent]:
            # Swap the current element with its parent if it's greater
            self.heap[index], self.heap[parent] = self.heap[parent], self.heap[index]
            self._heapify_up(parent)  # Recursively fix the heap

    def extract_max(self):
        """Remove and return the maximum value from the heap."""
        if len(self.heap) == 0:
            return None  # Return None if the heap is empty
        if len(self.heap) == 1:
            return self.heap.pop()  # If there's only one element, just remove it
        # Swap the root with the last element, then restore the heap property
        root = self.heap[0]
        self.heap[0] = self.heap.pop()
        self._heapify_down(0)  # Restore heap property after extraction
        return root

    def _heapify_down(self, index):
        """Move the element at the given index down to restore the heap property."""
        left_child = 2 * index + 1  # Left child index
        right_child = 2 * index + 2  # Right child index
        largest = index

        # Check if the left child is larger than the current node
        if left_child < len(self.heap) and self.heap[left_child] > self.heap[largest]:
            largest = left_child

        # Check if the right child is larger than the current node
        if right_child < len(self.heap) and self.heap[right_child] > self.heap[largest]:
            largest = right_child

        # If the largest is not the current node, swap and continue heapifying down
        if largest != index:
            self.heap[index], self.heap[largest] = self.heap[largest], self.heap[index]
            self._heapify_down(largest)

    def peek(self):
        """Return the maximum value (root) without removing it."""
        return self.heap[0] if self.heap else None

    def size(self):
        """Return the number of elements in the heap."""
        return len(self.heap)

    def display(self):
        """Display the current heap."""
        print(self.heap)

# Example usage:
heap = MaxHeap()
heap.insert(10)
heap.insert(20)
heap.insert(5)
heap.insert(30)
heap.insert(15)

heap.display()  # Output: [30, 15, 5, 10, 20]

print("Max value:", heap.extract_max())  # Output: 30
heap.display()  # Output: [20, 15, 5, 10]

print("Max value:", heap.peek())  # Output: 20
```

```python
class Graph:
    def __init__(self):
        self.graph = {}  # Dictionary to store graph as an adjacency list

    def add_vertex(self, vertex):
        """Add a vertex to the graph."""
        if vertex not in self.graph:
            self.graph[vertex] = []

    def add_edge(self, vertex1, vertex2, directed=False):
        """Add an edge between two vertices."""
        if vertex1 not in self.graph:
            self.add_vertex(vertex1)
        if vertex2 not in self.graph:
            self.add_vertex(vertex2)

        self.graph[vertex1].append(vertex2)

        if not directed:
            self.graph[vertex2].append(vertex1)  # For undirected graph, add reverse edge

    def display(self):
        """Display the adjacency list of the graph."""
        for vertex in self.graph:
            print(f"{vertex}: {self.graph[vertex]}")

    def bfs(self, start):
        """Breadth-First Search (BFS) starting from a given vertex."""
        visited = set()
        queue = [start]
        visited.add(start)

        while queue:
            vertex = queue.pop(0)
            print(vertex, end=" ")

            for neighbor in self.graph[vertex]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)

    def dfs(self, start):
        """Depth-First Search (DFS) starting from a given vertex."""
        visited = set()
        self._dfs_helper(start, visited)

    def _dfs_helper(self, vertex, visited):
        visited.add(vertex)
        print(vertex, end=" ")

        for neighbor in self.graph[vertex]:
            if neighbor not in visited:
                self._dfs_helper(neighbor, visited)

# Example usage
g = Graph()

# Adding vertices
g.add_vertex(1)
g.add_vertex(2)
g.add_vertex(3)

# Adding edges (undirected by default)
g.add_edge(1, 2)
g.add_edge(1, 3)
g.add_edge(2, 3)

# Display the graph
g.display()

# Perform BFS and DFS starting from vertex 1
print("\nBFS starting from vertex 1:")
g.bfs(1)

print("\nDFS starting from vertex 1:")
g.dfs(1)
```

```python
#trie

class TrieNode:
    """Represents a single node in the Trie."""
    def __init__(self):
        self.children = {}  # Dictionary to store children nodes
        self.is_end_of_word = False  # Flag to mark the end of a word

class Trie:
    """Represents the Trie data structure."""
    def __init__(self):
        self.root = TrieNode()  # Root node is an empty TrieNode

    def insert(self, word):
        """Insert a word into the Trie."""
        node = self.root
        for char in word:
            # If the character is not in the current node's children, add it
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_end_of_word = True  # Mark the end of the word

    def search(self, word):
        """Search for a word in the Trie."""
        node = self.root
        for char in word:
            if char not in node.children:
                return False  # Return False if the word is not found
            node = node.children[char]
        return node.is_end_of_word  # Return True if the word is found

    def starts_with(self, prefix):
        """Check if there's any word in the Trie that starts with the given prefix."""
        node = self.root
        for char in prefix:
            if char not in node.children:
                return False  # Return False if the prefix is not found
            node = node.children[char]
        return True  # Return True if the prefix exists in the Trie

# Example usage
trie = Trie()
trie.insert("apple")
trie.insert("app")
trie.insert("banana")

# Search for a word
print(trie.search("apple"))  # Output: True
print(trie.search("app"))    # Output: True
print(trie.search("appl"))   # Output: False

# Check for a prefix
print(trie.starts_with("ban"))  # Output: True
print(trie.starts_with("bat"))  # Output: False
```

```python
# union find

class UnionFind:
    def __init__(self, size):
        # Initialize parent and rank arrays
        self.parent = list(range(size))  # Each element is its own parent initially
        self.rank = [0] * size  # All elements have rank 0 initially

    def find(self, x):
        """Find the root of the element x with path compression."""
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]

    def union(self, x, y):
        """Union the sets containing x and y using union by rank."""
        rootX = self.find(x)
        rootY = self.find(y)

        if rootX != rootY:
            # Union by rank: attach the smaller tree under the larger one
            if self.rank[rootX] > self.rank[rootY]:
                self.parent[rootY] = rootX
            elif self.rank[rootX] < self.rank[rootY]:
                self.parent[rootX] = rootY
            else:
                self.parent[rootY] = rootX
                self.rank[rootX] += 1  # Increment rank if the trees are the same height

    def connected(self, x, y):
        """Check if x and y are in the same set."""
        return self.find(x) == self.find(y)

# Example usage
uf = UnionFind(5)

# Union some sets
uf.union(0, 1)
uf.union(1, 2)
uf.union(3, 4)

# Check connections
print(uf.connected(0, 2))  # Output: True
print(uf.connected(0, 3))  # Output: False

# Union sets again
uf.union(2, 3)

# Check again
print(uf.connected(0, 3))  # Output: True
```
